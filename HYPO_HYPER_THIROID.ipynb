```python
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
from google.colab import files

# Upload thermal image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Load original image (no aggressive resize)
img = Image.open(filename).convert("RGB")
pixels = np.array(img)

# Grayscale conversion
gray_img = np.uint8(np.mean(pixels, axis=2))

# High temperature threshold
high_temp_thresh = 150

# Smooth high temperature map
# Only keep values above threshold, scaled for intensity
high_temp_map = np.zeros_like(gray_img, dtype=np.uint8)
high_temp_map[gray_img >= high_temp_thresh] = gray_img[gray_img >= high_temp_thresh]

# Overlay high temperature map on original grayscale
overlay = np.stack([gray_img]*3, axis=2)  # convert to RGB
# Red highlight proportional to temperature above threshold
overlay[:,:,0] = np.maximum(overlay[:,:,0], high_temp_map)  # Red channel
overlay[:,:,1] = np.where(high_temp_map>0, overlay[:,:,1]//2, overlay[:,:,1])  # reduce green
overlay[:,:,2] = np.where(high_temp_map>0, overlay[:,:,2]//2, overlay[:,:,2])  # reduce blue

# Plot results
fig, ax = plt.subplots(1,3, figsize=(18,6))
ax[0].imshow(gray_img, cmap='gray')
ax[0].set_title("Grayscale Thermal Image")
ax[0].axis('off')

ax[1].imshow(high_temp_map, cmap='hot')
ax[1].set_title("High Temperature Map")
ax[1].axis('off')

ax[2].imshow(overlay)
ax[2].set_title("Overlay High Temp on Thermal Image")
ax[2].axis('off')

plt.tight_layout()
plt.show()
```
